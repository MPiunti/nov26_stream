```sql
-- ============================================================
-- 07 - Validate Bronze ingestion
-- ============================================================

SELECT
    COUNT(*) AS total_records,
    COUNT(DISTINCT source_offset) AS distinct_offsets,
    COUNT_IF(raw_event IS NULL) AS null_payloads,
    COUNT_IF(ingestion_timestamp IS NULL) AS null_ingestion_timestamps,
    COUNT_IF(source_timestamp IS NULL) AS null_source_timestamps
FROM workspace.test02_streaming.bronze_people_counter_events;

-- Inspect ingested records
SELECT
    source_topic,
    source_partition,
    source_offset,
    source_timestamp,
    ingestion_timestamp,
    raw_event
FROM workspace.test02_streaming.bronze_people_counter_events
ORDER BY source_offset;
```
